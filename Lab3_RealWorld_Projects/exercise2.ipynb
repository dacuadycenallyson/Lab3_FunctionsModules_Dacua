```python
# --- RECURSIVE FAULT TRACE SYSTEM ---

# Student-Specific Inputs
LAST_NAME = "DACUA"
SEED_NUM = 1
FAVORITE_ARTIST = "MYMP"

# Generate student-specific initial fault code
initial_fault_code = (len(LAST_NAME) * SEED_NUM * len(FAVORITE_ARTIST)) * 5  # 5 * 1 * 4 * 5 = 100

print("--- RECURSIVE FAULT TRACE SYSTEM ---")
print(f"Student: {LAST_NAME} | Seed: {SEED_NUM} | Artist: {FAVORITE_ARTIST}")
print(f"Generated Fault Data (Initial Code): {initial_fault_code}\n")

# Global counter for recursive calls
call_count = 0

def recursive_fault_trace(fault_code):
    global call_count
    call_count += 1
    
    # Base condition: stop when fault code drops to 0 or below
    if fault_code <= 0:
        print(f"[Base Condition Reached]: Fault code reached {fault_code}. Trace terminated.")
        return "STABLE / RESOLVED"
    
    print(f"Call {call_count}: Current Fault Code = {fault_code}")
    
    # Recursive step: reduce the fault code
    reduced_code = fault_code - 15
    return recursive_fault_trace(reduced_code)

print("--- RECURSIVE TRACE LOG ---")
final_result = recursive_fault_trace(initial_fault_code)

print("\n--- ASSESSMENT DATA ---")
print(f"Generated Fault Data: {initial_fault_code}")
print(f"Recursive Trace: Completed down to base condition")
print(f"Number of Recursive Calls: {call_count}")
print(f"Final Output / Result: {final_result}")
```
