```python
# --- EQUIPMENT DIAGNOSTIC SYSTEM ---

# Student-Specific Inputs
LAST_NAME = "DACUA"
SEED_NUM = 1
FAVORITE_ARTIST = "MYMP"

# Generate student-specific equipment readings
raw_readings = [SEED_NUM * 15, len(LAST_NAME) * 5, "INVALID_VAL", -10, len(FAVORITE_ARTIST) * 4]

print("--- EQUIPMENT DIAGNOSTIC SYSTEM ---")
print(f"Student: {LAST_NAME} | Seed: {SEED_NUM} | Artist: {FAVORITE_ARTIST}")
print(f"Generated Equipment Data (Raw): {raw_readings}")

def log_execution(func):
    def wrapper(*args, **kwargs):
        print("[DECORATOR LOG]: Executing diagnostic function...")
        result = func(*args, **kwargs)
        print("[DECORATOR LOG]: Execution completed successfully.")
        return result
    return wrapper

def validate_readings(readings):
    valid = []
    for r in readings:
        try:
            val = float(r)
            if val < 0:
                raise ValueError("Negative reading detected.")
            valid.append(val)
        except (ValueError, TypeError) as e:
            print(f"[Exception Handled]: Invalid entry '{r}' skipped. Reason: {e}")
    return valid

def calculate_metrics(valid_readings):
    if not valid_readings:
        return 0.0, 0.0
    total = sum(valid_readings)
    avg = total / len(valid_readings)
    return total, avg

@log_execution
def classify_equipment(avg_val):
    if avg_val > 50:
        return "OPTIMAL STATUS"
    elif avg_val > 20:
        return "MODERATE WARNING"
    else:
        return "CRITICAL FAULT"

valid_data = validate_readings(raw_readings)
total_sum, average_val = calculate_metrics(valid_data)
status = classify_equipment(average_val)

print("\n--- ASSESSMENT DATA ---")
print(f"Validation Results: {valid_data}")
print(f"Diagnostic Results: Total={total_sum}, Average={round(average_val, 2)}")
print(f"Final Output / Equipment Condition: {status}")
```
